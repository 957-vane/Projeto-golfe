```python
#Importar as bibliotecas necessárias 
import pandas as pd 
```

```python
#Ler ficheiro e ignorar as 6 primeiras linhas (deixando somente a tabela)
dados = pd.read_excel("Dados FQ solo.xlsx", skiprows= 6)

#Mostrar o ficheiro 
print(dados.head())
```

```python
#Inserir colunas ()
dados.insert(0, "raw_id", range(1, len(dados) + 1))

print(dados.head())
```

```python
#Remomear colunas 
dados.rename(columns={
    "Amostra": "sample_code",
    "Parcela": "parcel_name",
    "Data colheita": "sample_date",
    "Família Parâmetro": "parameter_name",
    "Resultado*": "parameter_value",
    "Unidades": "parameter_unit",
    "Interpretação": "interpretation",
    "Recomendação de fertilização/Comentário": "recommendations",
    "Método": "method"
}, inplace=True)

print(dados.info())
```

```python
#Remover colunas 
dados = dados.drop(columns= ["Cliente", "Cultura", "Localização", "Técnica" ])

print(dados.info())
```

```python
#Adicionar novas colunas 
dados.insert(10,"loaded_at", "")
dados.insert(11, "source_file", "raw_soil_fq.csv")

#Visualizar tabela 
print(dados.head())
```

```python
#Transformar dados em CSV utilizando o separador ;
dados.to_csv("raw_soil_fq.csv", sep=";", index=False)
```
