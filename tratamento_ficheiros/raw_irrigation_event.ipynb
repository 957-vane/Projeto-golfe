```python
#Importar bibliotecas necessárias 
import pandas as pd 
```

```python
#Importar o ficheiro 
caminho = "Exportestufa 13000 m2.xls"
dados = pd.read_excel(caminho, engine='xlrd')

#Mostrar dados 
print(dados.info())
```

```python
#Substituir quebras de linha nos nomes da coluna por espaço 
dados.columns = dados.columns.str.replace('\n', ' ', regex=False)

#Mostrar dados 
print(dados.info())
```

```python
#Renomear colunas 
dados.rename(columns={
    "Data": "event_date",
    "Hora de início": "start_time",
    "Hora de fim": "end_time",
    "Rest Time": "rest_time",
    "MCU": "mcu",
    "n°": "event_number",
    "Grupo": "group",
    "Numbert of irrigation": "irrigation_number",
    "Modo": "mude",
    "Ec instrução": "ec_instruction",
    "Ec medida": "ec_measured",
    "ph instrução": "ph_instruction",
    "ph medida": "ph_measured",
    "Tempo": "duration_text",
    "Volume": "volume_applied",
    "Volume de drenagem": "drainage_volume",
    "% de drenagem": "drainage_pct",
    "Readout Ec drain medida": "drain_ec_measured",
    "Readout pH drain medida": "drain_ph_measured",
}, inplace=True)

#Visualizar colunas 
print(dados.info())
```

```python
#Remover colunas 
dados = dados.drop(columns=[
    "Ecp instrução",
    "Ecp medida"
])

#Mostrar resultado 
print(dados.info())
```

```python
#Inserir nova coluna 

dados["source_file"] = "raw_irrigation_event.csv"

dados["loaded_at"] = None

dados.insert(0, "raw_id", range(1,len(dados) + 1))

#Mostrar resultado 

print(dados.info())
```

```python
print(dados.head())
```

```python
# Remove duplicatas e retorna um novo DataFrame
df_nulos = dados.dropna(subset=['start_time'])

print(df_nulos.head())
```

```python
#Transformar o ficheiro em csv já tratado 

df_nulos.to_csv("raw_irrigation_event.csv", sep=";", index=False)
```
